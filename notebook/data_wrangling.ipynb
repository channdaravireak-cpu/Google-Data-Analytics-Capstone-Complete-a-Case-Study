```python
import pandas as pd
df = pd.read_csv('data\\raw\\combined_data.csv')
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df['started_at'] = pd.to_datetime(df['started_at'])
df['ended_at'] = pd.to_datetime(df['ended_at'])
df['ride_length'] = df['ended_at'] - df['started_at']
```

```python
df['ride_length'] = df['ride_length'].apply(
    lambda x: f"{int(x.total_seconds() // 3600):02d}:{int((x.total_seconds() % 3600) // 60):02d}:{int(x.total_seconds() % 60):02d}"
)
```

```python
df['day_of_week'] = df['started_at'].dt.day_name()
```

```python
df.drop_duplicates(keep='first', inplace=True)
print(df.duplicated().sum())
```

```python
df['start_station_name'].isnull().sum()/len(df)*100
```

```python
df['end_station_name'].isnull().sum()/len(df)*100
```

```python
df['end_lat'].isnull().sum()/len(df)*100
```

```python
df.dropna(subset=['end_lat','end_lng'], inplace=True)
print(df[['end_lat','end_lng']].isnull().sum())
```

```python
df['start_station_name'] = df['start_station_name'].fillna(df['start_station_name'].mode()[0])
df['end_station_name'] = df['end_station_name'].fillna(df['end_station_name'].mode()[0])
df['end_station_id'] = df['end_station_id'].fillna(df['end_station_id'].mode()[0])
df['start_station_id'] = df['start_station_id'].fillna(df['start_station_id'].mode()[0])
df.isnull().sum()
```

```python
df.to_csv('data\\clean\\cleaned_data.csv', index=False)
```
